# Part I — Foundations

## 1. What Git Is and How to Think About It

Git is a **version control system**.  
Its job is to help you:

- track changes over time
- move safely between versions
- work on multiple lines of development
- collaborate with other people
- recover from mistakes

A good way to think about Git is this:

> Git records project history as a sequence of snapshots.

Each commit is a recorded snapshot with metadata such as:

- author
- date
- message
- parent commit(s)

### Why This Matters

If the learner does not understand that Git is fundamentally about **history and snapshots**, then later commands such as `log`, `branch`, `merge`, `rebase`, and `revert` feel random.


## 2. Working Tree, Staging Area, and Commit History

Before learning commands, you need the three-layer model.

### Working Tree
The actual files in your project folder.

### Staging Area
The set of changes prepared for the next commit.

### Commit History
The already-recorded snapshots stored by Git.

### Visual Model

```text
Working Tree  →  Staging Area  →  Commit History
   edit            git add          git commit
```

### Why This Matters

Many beginner Git mistakes happen because these layers are confused.

For example:

- `git diff` may compare working changes
- `git diff --staged` compares staged changes
- `git restore --staged` affects the index, not necessarily the file content


## 3. Repository Initialization with `git init -b main`

To create a new Git repository:

```bash
git init -b main
```

### What This Does

- creates a `.git/` directory
- initializes the repository
- creates the initial branch as `main`

### Why `-b main` Matters

Older setups often used `master`.  
Modern projects usually prefer `main`.

### Alternative Global Setting

```bash
git config --global init.defaultBranch main
```

Then plain `git init` will also use `main`.


````python
git init -b main
git config --global init.defaultBranch main
````

## 4. Basic Configuration with `git config`

Git configuration controls identity and behavior.

Common early settings:

```bash
git config --global user.name "Your Name"
git config --global user.email "you@example.com"
git config --global core.editor "code --wait"
git config --global color.ui auto
```

### Levels of Configuration

- `--system`
- `--global`
- `--local`

Priority is generally:

```text
local > global > system
```

### Why This Matters

Without correct config:

- commits may show the wrong author
- Git may open an editor you do not want
- behavior may differ across repositories


````python
git config --global user.name "Your Name"
git config --global user.email "you@example.com"
git config --global core.editor "code --wait"
git config --global color.ui auto
git config --list
````

## 5. Viewing Configuration Origins

A very useful debugging command is:

```bash
git config --list --show-origin
```

This shows:

- the config value
- the file where it came from

Example idea:

```text
file:/etc/gitconfig        core.editor=vim
file:/home/user/.gitconfig user.name=Ali
file:.git/config           core.repositoryformatversion=0
```

### Why This Matters

Sometimes the same setting exists in multiple places.  
This command helps you see which file is responsible.


````python
git config --list --show-origin
````

## 6. Creating Aliases

Git aliases let you define shorter names for longer commands.

Example:

```bash
git config --global alias.co checkout
git config --global alias.br branch
git config --global alias.cm commit
git config --global alias.st status
git config --global alias.lg "log --oneline --graph --all"
```

Then you can use:

```bash
git co main
git br
git cm -m "Fix typo"
git st
git lg
```

### Why This Matters

Aliases reduce repetition and speed up everyday work.

### Important Note

Aliases are Git-specific.  
They do not become global shell commands.


````python
git config --global alias.co checkout
git config --global alias.br branch
git config --global alias.cm commit
git config --global alias.st status
git config --global alias.lg "log --oneline --graph --all" 
````

## 7. Core Snapshot Commands: `add`, `commit`, `log`, `diff`

These commands form the basic local Git loop.

### Stage changes
```bash
git add .
```

### Commit staged changes
```bash
git commit -m "Add new feature"
```

### View history
```bash
git log
git log --oneline
git log --oneline --graph --all
```

### Compare changes
```bash
git diff
git diff --staged
git diff HEAD
```

### Why This Matters

This is the minimum working loop of Git:

```text
edit → add → commit → inspect
```

If this loop is solid, the rest of Git becomes much easier to understand.


````python
git add .
git commit -m "Add new feature"

git log --oneline --graph --all

git diff
git diff --staged
git diff HEAD
````

# Part II — Branching and Local Workflow

## 8. Why Branches Matter

A branch gives you a separate line of development.

Instead of putting every change directly onto `main`, you can isolate work.

### Why Branches Are Important

Branches let you:

- keep `main` stable
- work on features or fixes separately
- experiment without damaging the main line
- create clean Pull Requests later
- integrate work step by step

### Core Idea

A branch is not a second full copy of the project.  
It is a movable pointer into commit history.


## 9. `git branch` and `git switch`

### List branches
```bash
git branch
```

### Create a branch
```bash
git branch feature/login
```

### Switch to another branch
```bash
git switch main
```

### Create and switch in one step
```bash
git switch -c feature/login
```

### Why This Matters

Branching should be understood **before** Fork and Pull Request workflows, because PRs are usually built from branches.


````python
git branch
git switch -c feature/login
git switch main
````

## 10. A Clean Everyday Local Workflow

A clean local branch workflow often looks like this:

```text
main → create branch → make changes → commit → switch back when needed
```

Example:

```bash
git switch main
git pull
git switch -c fix/readme-typo
git add README.md
git commit -m "Fix typo in README"
```

### Key Takeaway

Before learning collaboration on GitHub, a learner should already be comfortable with:

- creating a branch
- switching branches
- committing on a branch
- understanding that work stays isolated until integrated


````python
git switch main
git pull
git switch -c fix/readme-typo
git add README.md
git commit -m "Fix typo in README" 
````

# Part III — Remotes, GitHub, and SSH

## 11. Why Remote Repositories Matter

A local repository is enough for solo experimentation, but collaboration requires a shared remote repository.

GitHub provides that shared remote space.

### Why Remotes Matter

They allow you to:

- back up your project online
- collaborate with teammates
- open Pull Requests
- fetch and push shared history

### Important Distinction

- local Git = history on your machine
- remote GitHub repo = shared history online


## 12. `clone`, `origin`, `push`, and `pull`

### Clone a remote repository
```bash
git clone git@github.com:OWNER/PROJECT.git
```

### What `origin` Usually Means

After cloning, Git usually names the default remote:

```text
origin
```

This is simply the default name for the remote source you cloned from.

### Push
```bash
git push origin main
```

### Pull
```bash
git pull
```

### Why This Matters

A lot of GitHub confusion disappears once the learner understands:

- `origin` is just a remote name
- `push` sends local commits
- `pull` brings remote updates


````python
git clone git@github.com:OWNER/PROJECT.git
git remote -v
git push origin main
git pull
````

## 13. HTTPS vs SSH

There are two common ways to connect Git to GitHub.

### HTTPS
Example:
```bash
git clone https://github.com/user/repo.git
```

### SSH
Example:
```bash
git clone git@github.com:user/repo.git
```

### General Guidance

- HTTPS is usually easier to start with
- SSH is often more convenient long-term, especially on Linux and macOS
- SSH avoids repeatedly typing credentials in many workflows

### Why This Matters

SSH is not a separate internet.  
It is a secure communication method that works over the network.


## 14. Creating an SSH Key

A common command is:

```bash
ssh-keygen -t ed25519 -C "your_email@example.com"
```

### What This Creates

Usually two files under `~/.ssh/`:

- private key: `id_ed25519`
- public key: `id_ed25519.pub`

### Important Rule

- private key stays on your machine
- public key is what you give to GitHub


````python
ssh-keygen -t ed25519 -C "your_email@example.com" 
````

## 15. How GitHub Verifies Your Private Key

A very important concept:

You do **not** copy your private key into GitHub.

Instead:

1. you generate a key pair locally
2. you upload only the **public key** to GitHub
3. when connecting, GitHub sends a challenge
4. your machine proves identity by signing that challenge with the **private key**
5. GitHub verifies the result using the stored **public key**

### Key Takeaway

GitHub never needs your private key file.


## 16. Testing SSH with GitHub

To test SSH authentication:

```bash
ssh -T git@github.com
```

### What `-T` Means

It disables interactive shell access and only tests authentication.

### Expected Successful Output

```text
Hi username! You've successfully authenticated, but GitHub does not provide shell access.
```

### Do You Need to Run This Every Time?

No.  
This is mainly a setup or debugging command.

Once SSH is configured correctly, normal Git commands will use it automatically.


````python
ssh -T git@github.com
````

# Part IV — GitHub Collaboration Models

## 17. Two Main Collaboration Models

There are two major GitHub collaboration models:

### 1. Fork Workflow
Used when you do **not** have write access to the main repository.

### 2. Contributor Workflow
Used when you **do** have write access to the main repository.

### Why This Section Order Matters

These two workflows should be taught together because they solve the same collaboration problem in two different permission contexts.


## 18. Fork Workflow

A fork is a copy of a repository under your own GitHub account.

### Typical Fork Workflow

```text
Fork → Clone your fork → Add upstream → Create branch → Commit → Push to fork → Pull Request
```

### Vocabulary

- `origin` = your fork
- `upstream` = the original repository

### Why Use It?

Use a fork when:

- you do not have write access
- you are contributing to open source
- you want a safe personal copy on GitHub

### Important Note

Forking gives you project history, but it does **not** automatically make you a contributor to the original repository.  
Usually, contribution becomes official when a PR is accepted and merged upstream.


````python
git clone git@github.com:YOUR_USERNAME/PROJECT.git
cd PROJECT
git remote add upstream git@github.com:ORIGINAL_OWNER/PROJECT.git
git switch -c my-feature
git add .
git commit -m "Describe your change"
git push origin my-feature
````

## 19. Contributor Workflow (Without Fork)

If you already have write access, a fork is often unnecessary.

### Typical Contributor Workflow

```text
Clone main repository → Create branch → Commit → Push branch to same repo → Pull Request
```

### Key Differences from Fork Workflow

- `origin` points directly to the shared repository
- there may be no need for `upstream`
- the branch usually lives in the main repo itself

### Important Note

Having write access does not mean you should push directly to `main`.  
Professional teams often still require:

- branches
- Pull Requests
- review
- CI checks
- protected branch rules


````python
git clone git@github.com:ORGANIZATION/PROJECT.git
cd PROJECT
git switch main
git pull
git switch -c fix/login-timeout
git add .
git commit -m "Fix login timeout handling"
git push origin fix/login-timeout
````

## 20. Pull Requests as a Review Mechanism

A Pull Request is not only a permission tool.  
It is also a collaboration and review mechanism.

### A PR Usually Provides

- code review
- discussion
- CI validation
- visibility into why the change exists
- a controlled path into `main`

### Two PR Shapes

#### Fork PR
```text
your-fork:branch → upstream:main
```

#### Same-Repository PR
```text
shared-repo:branch → shared-repo:main
```

### Key Takeaway

The branch source differs, but the review purpose is similar.


# Part V — Integration

## 21. Why Integration Is Hard

As soon as multiple branches exist, integration becomes necessary.

Typical reasons:

- your feature branch must be combined with `main`
- the upstream project moved while you were working
- your PR is outdated
- two people changed overlapping code

This is where merge, rebase, and conflict resolution matter.


## 22. Merge

Merge combines histories.

Example:

```bash
git switch main
git merge feature/login
```

### Characteristics of Merge

- usually creates a merge commit
- preserves branch structure
- does not rewrite existing commits

### Why Teams Like It

It is often safer for shared collaboration because it preserves visible history.


````python
git switch main
git merge feature/login
````

## 23. Rebase

Rebase replays commits on top of a new base.

Example:

```bash
git switch feature/login
git fetch origin
git rebase origin/main
```

### Characteristics of Rebase

- rewrites commit history
- changes commit hashes
- often creates a cleaner linear history

### Important Rule

Do not casually rebase history that other people are already using.


````python
git switch feature/login
git fetch origin
git rebase origin/main
````

## 24. Merge vs Rebase

### Merge
- preserves branch topology
- safer for shared history
- may create noisier logs

### Rebase
- produces linear history
- cleaner visual result
- rewrites history

### Simple Mental Model

- merge = bring histories together
- rebase = move my branch onto a newer base


## 25. Conflict Resolution

A conflict happens when Git cannot automatically decide how to combine overlapping changes.

### Common Conflict Markers

```text
<<<<<<< HEAD
current branch content
=======
incoming branch content
>>>>>>> feature/login
```

### Merge Conflict Flow

```bash
git switch main
git merge feature/login
git status
# edit conflicted files
git add path/to/file
git commit
```

### Rebase Conflict Flow

```bash
git switch feature/login
git rebase origin/main
git status
# edit conflicted files
git add path/to/file
git rebase --continue
```

### Useful Abort Commands

```bash
git merge --abort
git rebase --abort
```

### Key Takeaway

Conflict resolution is not only a Git skill.  
It is a code understanding skill.


````python
git merge --abort
git rebase --abort
git rebase --continue
````

# Part VI — Undo, Recovery, and Selective Reuse

## 26. Choosing the Right Undo Tool

Before using an undo-related command, ask:

### Question 1
Do I want to **rewrite history** or **preserve history**?

### Question 2
Am I changing **file state**, **staging state**, or **commit history**?

### Question 3
Has this already been shared with others?

This mental model helps you choose the right command.


## 27. `reset`

`git reset` moves the current branch pointer and may also affect:

- the staging area
- the working tree

### Common Modes

#### Soft
```bash
git reset --soft HEAD~1
```
Moves back one commit, keeps changes staged.

#### Mixed
```bash
git reset --mixed HEAD~1
```
Moves back one commit, keeps changes unstaged in the working tree.

#### Hard
```bash
git reset --hard HEAD~1
```
Moves back one commit and discards local working changes.

### Best Use Case

`reset` is usually best for **local cleanup before sharing**.


````python
git reset --soft HEAD~1
git reset --mixed HEAD~1
git reset --hard HEAD~1
````

## 28. `revert`

`git revert` creates a new commit that undoes an older commit.

Example:

```bash
git revert abc1234
```

### Why This Matters

Unlike `reset`, `revert` does **not** remove the original commit from history.

### Best Use Case

Use `revert` when a bad commit is already shared and you want a safe audit trail.


````python
git revert abc1234
````

## 29. `restore`

`git restore` is mainly about file-level state.

### Discard local changes in one file
```bash
git restore app.py
```

### Unstage one file
```bash
git restore --staged app.py
```

### Restore from a specific commit
```bash
git restore --source=abc1234 README.md
```

### Best Use Case

Use `restore` when your goal is to fix file state without doing branch history surgery.


````python
git restore app.py
git restore --staged app.py
git restore --source=abc1234 README.md
````

## 30. `commit --amend`

Amend replaces the most recent commit.

### Fix the latest message
```bash
git commit --amend -m "Better commit message"
```

### Add a forgotten file to the last commit
```bash
git add forgotten_file.py
git commit --amend
```

### Important Note

Amend rewrites the last commit, so the hash changes.

### Best Use Case

Use amend when the **latest local commit** needs a small correction.


````python
git commit --amend -m "Better commit message"
git add forgotten_file.py
git commit --amend
````

## 31. `cherry-pick`

Cherry-pick applies one specific commit onto your current branch.

Example:

```bash
git cherry-pick abc1234
```

### Best Use Cases

- backporting a fix
- reusing one commit without merging an entire branch
- applying a bug fix to a release branch

Example:

```bash
git switch release/1.2
git cherry-pick abc1234
```

### Important Note

Cherry-pick can also produce conflicts, and then you may need:

```bash
git cherry-pick --continue
git cherry-pick --abort
```


````python
git switch release/1.2
git cherry-pick abc1234
git cherry-pick --continue
git cherry-pick --abort
````

# Part VII — Final Review

## 32. Recommended Learning Order Recap

A strong Git learning order is:

```text
Foundations
→ Branching
→ Remotes and GitHub
→ SSH
→ Fork and Contributor workflows
→ Pull Requests
→ Merge / Rebase / Conflicts
→ Reset / Revert / Restore / Amend / Cherry-pick
```

### Why This Order Works

It introduces each concept after its prerequisites.

For example:

- branches before PRs
- remotes before SSH
- collaboration models together
- integration before advanced recovery

When these are taught in the right order, Git feels much more coherent and much less intimidating.
